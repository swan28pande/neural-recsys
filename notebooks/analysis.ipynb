# Neural Collaborative Filtering - Analysis

This notebook provides:
1. Exploratory Data Analysis (EDA) of the MovieLens-1M dataset
2. Training results visualization
3. Model comparison and ablation studies

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib

matplotlib.rcParams['figure.figsize'] = (10, 6)
matplotlib.rcParams['font.size'] = 12
plt.style.use('seaborn-v0_8-whitegrid')

%matplotlib inline
```

## 1. Load MovieLens-1M Data

```python
# Load raw ratings
ratings = pd.read_csv(
    '../data/ml-1m/ratings.dat', sep='::', engine='python',
    names=['user_id', 'item_id', 'rating', 'timestamp'],
    encoding='latin-1'
)

print(f'Ratings: {len(ratings):,}')
print(f'Users:   {ratings.user_id.nunique():,}')
print(f'Items:   {ratings.item_id.nunique():,}')
print(f'Density: {len(ratings) / (ratings.user_id.nunique() * ratings.item_id.nunique()) * 100:.2f}%')
ratings.head()
```

## 2. Rating Distribution

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Rating distribution
ratings['rating'].value_counts().sort_index().plot.bar(ax=axes[0], color='steelblue')
axes[0].set_title('Rating Distribution')
axes[0].set_xlabel('Rating')
axes[0].set_ylabel('Count')

# User activity distribution
user_counts = ratings.groupby('user_id').size()
axes[1].hist(user_counts, bins=50, color='coral', edgecolor='white')
axes[1].set_title('User Activity Distribution')
axes[1].set_xlabel('Number of Ratings')
axes[1].set_ylabel('Number of Users')
axes[1].set_yscale('log')

# Item popularity distribution
item_counts = ratings.groupby('item_id').size()
axes[2].hist(item_counts, bins=50, color='seagreen', edgecolor='white')
axes[2].set_title('Item Popularity Distribution')
axes[2].set_xlabel('Number of Ratings')
axes[2].set_ylabel('Number of Items')
axes[2].set_yscale('log')

plt.tight_layout()
plt.show()
```
